```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
# Data source: 
# One of the first and fundamental question in data science tasks, is how trusted and accurate is our data source
# In this project, we are using a customers sample dataset for learning purposes.

data = pd.read_csv("churn.csv")
# The column contract does not 
data["TypeOfContract"] = data["Contract"] # Giving the column a more significant name.
data = data.drop("Contract", axis=1)
# data.info()
data


# Now, our data is ready to move to the preprocessing stage 

```

```python
# Data integrity assessment :

# SeniorCitizen is an encoded categorical variable : Yes -> 1 , No -> 0
# All data types seem to be accurate and relevant to the associated features
# Some categorical features might need to be encoded for some other analysis such as correlation 
# Negative values in monthly and total charges are suspiscious, however this might be due to unpaid bills.

# ----------------------- Understanding what the grain represents, what one columns represents from a business perspective -------------

# The dataset is presented in a single table, so there are no relatioship to consider in this scenario.
# Each customer had a unique identifier. Each customer have personal characteristics (Gender, Seniority)
# As well as some business associated characteristics (Partner, Dependents, tenure, Services, Payment method, Charges, Type of contract)
# and a target feature (Churn) we might try to predict for new customers using some predictive modeling in later stages of this program
# 
```

```python
# Data Cleaning :

# Checking null values and assess the relevance of potential mitigation strategies (Deleting, Imputation, statistical methods, ..) :
Columns_with_Nulls = data.columns[data.isnull().any()].tolist()
print(Columns_with_Nulls)  

# First reflexion : does null values communicate a hidden information or simply represent missing or unavailable values.


# data["SeniorCitizen"] = data["SeniorCitizen"].replace({0: "No", 1: "Yes"})
# pd.set_option("display.max_rows", None)
Incomplete_rows = data[data.isnull().any(axis=1)]
#Incomplete_rows

Both_Missing = data[data[["MonthlyCharges", "TotalCharges"]].isnull().all(axis=1)]


# Only 6 rows have both columns of monthly charges and total charges missing (The common traits among all of them is that they are non senior customers and they all use phone service)
# We can delete those since they are a minor proportion of the dataset and have both information missing.

data.dropna(subset=["MonthlyCharges", "TotalCharges"], how="all", inplace=True)

# Also, it is very important to understand why these values are missing from a business perspective : (bugs in pipelines, or these missing values can have an accurate business interpretation)
# for other missing values where only one of the two is not present in the dataset, we can use some imputation technique to calculate total charges based on monthly charges and tenure
# and vice-versa, however this is a mitigation approach only as it may not consider the specific circumstances of each client. This limitation should be acknowledge and mentioned in our business intepretations
# Handle the case where tenure = 0 separately ;

data["TotalCharges"] = data["TotalCharges"].fillna(data["MonthlyCharges"] * data["tenure"])
data["MonthlyCharges"] = data["MonthlyCharges"].fillna(data["TotalCharges"].div(data["tenure"]).where(data["tenure"] > 0))

# Dropping the remaining missing values in MonthlyCharges column where tenure is null
data = data.dropna(subset=["MonthlyCharges"])


# Checking the data for duplicated values:
# a duplicate must be defined through the representation of our data, in our case a duplicate is a row where the exact same information for all columns are present more than once.

NumberOfduplicates = data.duplicated().sum()
# print(NumberOfduplicates) -> Our data contains 31 duplicates, in a real world case (we should raise the question about why those duplicates exist, why they were accepted and why our database rules did not prevent duplicated rows.)
duplicates = data[data.duplicated()]
data.drop_duplicates(inplace=True)

data.info()
```

```python
data["z_score_MonthlyCharges"] = stats.zscore(data["MonthlyCharges"])
outliers = data[data["z_score_MonthlyCharges"].abs() > 3]
# outliers.info()
data = data[data["z_score_MonthlyCharges"].abs() <= 3]
data.info()

# 24 outliers detected -> suspiscious values not necessarly invalid ones.
```

```python
# Manual Compute of z-score method -> because when column still have NaN values, they propagate and the z_score column return NaN for all values .
# This happens because the column still have 2 NaN values where the tenure is null

mean = data["MonthlyCharges"].mean()
std = data["MonthlyCharges"].std()

data["z_manual"] = (data["MonthlyCharges"] - mean) / std
outliers = data[data["z_manual"] > 3]
print(outliers)
```

```python
# IQR method : 

Q1 = data["TotalCharges"].quantile(0.25)
Q2 = data["TotalCharges"].quantile(0.50)
Q3 = data["TotalCharges"].quantile(0.75)

IQR = Q3 - Q2

upper_limit = Q3 + 1.5*IQR
lower_limit = Q1 - 1.5*IQR

outliers = data[(data["TotalCharges"] >= upper_limit) | (data["TotalCharges"] <= lower_limit)]
#outliers.info()
data = data[(data["TotalCharges"] <= upper_limit) & (data["TotalCharges"] >= lower_limit)]
data.info()

# 426 outliers 
```

```python
cols = ["InternetService", "PaymentMethod", "TypeOfContract", "PhoneService"]

fig, axes = plt.subplots(2, 2, figsize=(16, 10))
axes = axes.flatten()

for col, ax in zip(cols, axes):
    pd.crosstab(data[col], data["SeniorCitizen"]).plot(
        kind="bar",
        ax=ax,
        color=["red", "black"],
        edgecolor="black"
    )

    ax.set_title(col, fontweight="bold")
    ax.set_xlabel("")
    ax.set_ylabel("Count", fontweight="bold")
    ax.legend(["Non-Senior", "Senior"])
    ax.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python

cat_vars = [
    "gender",
    "SeniorCitizen",
    "Partner",
    "Dependents",
    "PhoneService",
    "InternetService",
    "PaymentMethod"
]

fig, axes = plt.subplots(4, 2, figsize=(14, 16))
axes = axes.ravel()

for ax, var in zip(axes, cat_vars):
    cross_pct = pd.crosstab(data[var], data["Churn"], normalize="index") * 100

    sns.heatmap(
        cross_pct,
        annot=cross_pct.round(1).astype(str) + "%",
        fmt="",
        cmap="Greys",
        linewidths=0.5,
        ax=ax
    )

    ax.set_title(f"Churn by {var}")
    ax.set_xlabel("Churn")
    ax.set_ylabel(var)

axes[-1].remove()

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import chi2_contingency

for var in cat_vars:
    cross = pd.crosstab(data[var], data["Churn"])
    
    chi2, p, dof, expected = chi2_contingency(cross)
    
    print(f"{var}")
    print(f"p-value = {p:.4f}")
    
    if p < 0.05:
        print("Significant association with Churn")
    else:
        print("No significant association with Churn")
    
    print("-" * 40)
```
